# Day 3 Homework — Error Handling and Input Validation

**Objective:** Demonstrate how invalid inputs are rejected using `ValueError` and how exceptions are handled using `try` and `except`.

**Predictions:**
- `None`: Rejected because it is not a string.
- Blank spaces: Rejected because the input is empty after stripping whitespace.
- 501 characters: Rejected because the input exceeds the 500-character limit.
- Valid question: Accepted after removing surrounding whitespace.

```python

# Day 3 Homework — Error Handling and Input Validation
import json


def validate_public_question(question):
    if not isinstance(question, str) or not question.strip():
        raise ValueError('Enter a non-empty question string.')
    if len(question) > 500:
        raise ValueError('Use at most 500 characters.')
    return question.strip()


def run_error_examples():
    cases = [
        (None, 'Enter a non-empty question string.'),
        ('   ', 'Enter a non-empty question string.'),
        ('x' * 501, 'Use at most 500 characters.')
    ]

    rows = []
    for value, expected in cases:
        try:
            validate_public_question(value)
        except ValueError as error:
            assert str(error) == expected
            rows.append({
                'error_type': type(error).__name__,
                'message': str(error),
                'expected_rejection': True
            })
        else:
            raise AssertionError('Invalid input was accepted.')

    assert validate_public_question(
        ' How can I check my application status? '
    ) == 'How can I check my application status?'

    result = {
        'intentional_caught_errors': rows,
        'corrected_input': 'accepted',
        'checks': 'passed'
    }

    print(json.dumps(result, indent=2))
    return result


run_error_examples()
```

### Explanation

- **`raise`:** Raises a `ValueError` when the input is invalid.
- **`try`:** Executes the validation function to check whether the input is accepted.
- **`except`:** Catches the expected `ValueError` and verifies its message.

**Why is a caught rejection a successful test?**

A caught rejection is considered successful because the validation function correctly identifies invalid input and prevents it from being accepted.

**Corrected Input:** `How can I check my application status?`

The corrected question is valid because it is a non-empty string containing fewer than 500 characters.

**Result:** All three intentional errors were caught successfully, and the corrected input was accepted.
